```python
def generate_customers(NUM_CUSTOMERS):
    """
    Generates realistic customer master data
    for ShopSphere.
    """

    customers = []

    for customer_id in range(1, NUM_CUSTOMERS + 1):

        # Customer Name
        name = fake.name()

        # Customer Gender
        gender = random.choice(["Male", "Female"])

        # Customer Age
        age = random.randint(18, 65)

        # Select Region
        region = random.choice(regions)

        # Select City belonging to selected Region
        city = random.choice(cities[region])

        # Contact Details
        email = fake.email()
        phone = fake.phone_number()

        # Customer Join Date
        join_date = fake.date_between(
            start_date="-5y",
            end_date="today"
        )

        # Create Customer Dictionary
        customer = {
            "CustomerID": customer_id,
            "CustomerName": name,
            "Gender": gender,
            "Age": age,
            "Email": email,
            "Phone": phone,
            "Region": region,
            "City": city,
            "JoinDate": join_date
        }

        # Store Customer
        customers.append(customer)

    return customers

customers = generate_customers(NUM_CUSTOMERS)

customers_df = pd.DataFrame(customers)

print(customers_df.head())

print("\n")

print(customers_df.info())

print("\nShape:", customers_df.shape)

customers_df.to_csv(
    "../data/raw/customers.csv",
    index=False
)
```

```python
# ==========================================================
# PRODUCT MASTER DATA
# ==========================================================
import pandas as pd

product_catalog = {

    "Electronics": {
        "Apple": [
            "iPhone 16",
            "MacBook Air M3",
            "AirPods Pro"
        ],
        "Samsung": [
            "Galaxy S25",
            "Galaxy Tab",
            "Galaxy Buds"
        ],
        "Sony": [
            "Bravia TV",
            "WH-1000XM5 Headphones"
        ],
        "HP": [
            "HP Pavilion Laptop"
        ]
    },

    "Fashion": {
        "Nike": [
            "Running Shoes",
            "Sports T-Shirt"
        ],
        "Adidas": [
            "Sneakers",
            "Track Pants"
        ],
        "Levis": [
            "Slim Jeans"
        ]
    },

    "Beauty": {
        "Lakme": [
            "Lipstick",
            "Face Wash"
        ],
        "L'Oreal": [
            "Shampoo",
            "Serum"
        ]
    },

    "Home": {
        "Philips": [
            "Mixer Grinder"
        ],
        "Milton": [
            "Water Bottle"
        ],
        "Prestige": [
            "Cookware Set"
        ]
    },

    "Grocery": {
        "Aashirvaad": [
            "Atta"
        ],
        "Fortune": [
            "Cooking Oil"
        ],
        "Tata": [
            "Tea"
        ]
    }
}

price_ranges = {
    "Electronics": (8000, 150000),
    "Fashion": (500, 6000),
    "Beauty": (200, 3000),
    "Home": (500, 15000),
    "Grocery": (50, 1000)
}


# ==========================================================
# GENERATE PRODUCTS
# ==========================================================

def generate_products(num_products):

    products = []

    for product_id in range(1, num_products + 1):

        category = random.choice(list(product_catalog.keys()))

        brand = random.choice(list(product_catalog[category].keys()))

        product_name = random.choice(product_catalog[category][brand])

        min_price, max_price = price_ranges[category]

        mrp = random.randint(min_price, max_price)

        discount_percent = random.randint(5, 35)

        selling_price = round(
            mrp * (1 - discount_percent / 100),
            2
        )

        cost_price = round(
            selling_price * random.uniform(0.60, 0.80),
            2
        )

        product = {
            "ProductID": product_id,
            "ProductName": product_name,
            "Category": category,
            "Brand": brand,
            "MRP": mrp,
            "DiscountPercent": discount_percent,
            "SellingPrice": selling_price,
            "CostPrice": cost_price
        }

        products.append(product)

    return products


# ==========================================================
# GENERATE DATA
# ==========================================================

products = generate_products(NUM_PRODUCTS)

products_df = pd.DataFrame(products)

print(products_df.head())

print("\nShape:", products_df.shape)

print("\nColumns:")
print(products_df.columns)

print("\nInfo:")
print(products_df.info())

# ==========================================================
# SAVE CSV
# ==========================================================

products_df.to_csv(
    "../data/raw/products.csv",
    index=False
)

print("\n✅ products.csv saved successfully!")
```

```python
# ==========================================================
# ORDERS DATA GENERATION
# ==========================================================

NUM_ORDERS = 10000

payment_methods = [
    "UPI",
    "Credit Card",
    "Debit Card",
    "Net Banking",
    "Cash on Delivery"
]

order_statuses = [
    "Delivered",
    "Delivered",
    "Delivered",
    "Delivered",
    "Delivered",
    "Delivered",
    "Delivered",
    "Cancelled",
    "Returned"
]


def generate_orders(num_orders):

    orders = []

    for order_id in range(1, num_orders + 1):

        customer = customers_df.sample(1).iloc[0]

        product = products_df.sample(1).iloc[0]

        quantity = random.randint(1, 5)

        order_date = fake.date_between(
            start_date="-2y",
            end_date="today"
        )

        payment_method = random.choice(payment_methods)

        order_status = random.choice(order_statuses)

        total_amount = round(
            product["SellingPrice"] * quantity,
            2
        )

        order = {

            "OrderID": order_id,

            "CustomerID": customer["CustomerID"],

            "ProductID": product["ProductID"],

            "Quantity": quantity,

            "OrderDate": order_date,

            "PaymentMethod": payment_method,

            "OrderStatus": order_status,

            "TotalAmount": total_amount

        }

        orders.append(order)

    return orders


# ==========================================================
# GENERATE ORDERS
# ==========================================================

orders = generate_orders(NUM_ORDERS)

orders_df = pd.DataFrame(orders)

print(orders_df.head())

print("\nShape :", orders_df.shape)

print("\nInfo")
print(orders_df.info())

print("\nNull Values")
print(orders_df.isnull().sum())

print("\nOrder Status Distribution")
print(orders_df["OrderStatus"].value_counts())

print("\nPayment Method Distribution")
print(orders_df["PaymentMethod"].value_counts())


# ==========================================================
# SAVE CSV
# ==========================================================

import os

os.makedirs("../Data/raw", exist_ok=True)

orders_df.to_csv(
    "../data/raw/orders.csv",
    index=False
)

print("\n✅ orders.csv saved successfully!")
```

```python
# ==========================================================
# PAYMENTS DATA GENERATION
# ==========================================================

payment_statuses = [
    "Success",
    "Success",
    "Success",
    "Success",
    "Success",
    "Success",
    "Failed",
    "Refunded"
]

payments = []

for _, order in orders_df.iterrows():

    payment_id = f"PAY{100000 + order['OrderID']}"

    transaction_id = fake.uuid4()[:16].upper()

    payment_status = random.choice(payment_statuses)

    if payment_status == "Success":
        payment_date = order["OrderDate"] + timedelta(days=random.randint(0, 2))
    elif payment_status == "Refunded":
        payment_date = order["OrderDate"] + timedelta(days=random.randint(5, 20))
    else:
        payment_date = order["OrderDate"]

    payment = {

        "PaymentID": payment_id,

        "OrderID": order["OrderID"],

        "TransactionID": transaction_id,

        "PaymentMethod": order["PaymentMethod"],

        "AmountPaid": order["TotalAmount"],

        "PaymentDate": payment_date,

        "PaymentStatus": payment_status

    }

    payments.append(payment)

payments_df = pd.DataFrame(payments)

print(payments_df.head())

print("\nShape :", payments_df.shape)

print("\nInfo")
print(payments_df.info())

print("\nNull Values")
print(payments_df.isnull().sum())

print("\nPayment Status Distribution")
print(payments_df["PaymentStatus"].value_counts())

# ==========================================================
# SAVE CSV
# ==========================================================

import os

os.makedirs("../Data/raw", exist_ok=True)

payments_df.to_csv(
    "../data/raw/payments.csv",
    index=False
)

print("\n✅ payments.csv saved successfully!")
```

```python
# ==========================================================
# REVIEWS DATA GENERATION
# ==========================================================

review_titles = [
    "Excellent Product",
    "Highly Recommended",
    "Very Good",
    "Worth the Money",
    "Satisfied",
    "Average Product",
    "Could Be Better",
    "Not Worth It",
    "Poor Quality",
    "Amazing Purchase"
]

positive_reviews = [
    "Excellent quality and fast delivery.",
    "Loved the product. Highly recommended.",
    "Value for money. Will buy again.",
    "Very satisfied with the purchase.",
    "Packaging was good and product is amazing.",
    "Exactly as described.",
    "Great experience with this product.",
    "Superb quality and easy to use."
]

neutral_reviews = [
    "Product is okay.",
    "Average quality.",
    "Expected a little better.",
    "Works as intended.",
    "Delivery was on time."
]

negative_reviews = [
    "Very disappointed.",
    "Poor quality product.",
    "Not worth the price.",
    "Received a defective item.",
    "Will not recommend.",
    "Bad packaging.",
    "Stopped working after few days."
]

reviews = []

review_id = 1

for _, order in orders_df.iterrows():

    # Only delivered orders receive reviews
    if order["OrderStatus"] != "Delivered":
        continue

    # Around 70% of delivered orders get a review
    if random.random() > 0.70:
        continue

    rating = random.choices(
        [5, 4, 3, 2, 1],
        weights=[35, 30, 20, 10, 5],
        k=1
    )[0]

    if rating >= 4:
        review_text = random.choice(positive_reviews)
    elif rating == 3:
        review_text = random.choice(neutral_reviews)
    else:
        review_text = random.choice(negative_reviews)

    review_date = order["OrderDate"] + timedelta(
        days=random.randint(3, 15)
    )

    review = {

        "ReviewID": review_id,

        "OrderID": order["OrderID"],

        "CustomerID": order["CustomerID"],

        "ProductID": order["ProductID"],

        "Rating": rating,

        "ReviewTitle": random.choice(review_titles),

        "ReviewText": review_text,

        "ReviewDate": review_date

    }

    reviews.append(review)

    review_id += 1

reviews_df = pd.DataFrame(reviews)

print(reviews_df.head())

print("\nShape :", reviews_df.shape)

print("\nInfo")
print(reviews_df.info())

print("\nNull Values")
print(reviews_df.isnull().sum())

print("\nRating Distribution")
print(reviews_df["Rating"].value_counts().sort_index())

# ==========================================================
# SAVE CSV
# ==========================================================

import os

os.makedirs("../Data/raw", exist_ok=True)

reviews_df.to_csv(
    "../data/raw/reviews.csv",
    index=False
)

print("\n✅ reviews.csv saved successfully!")
```

```python
# ==========================================================
# RETURNS DATA GENERATION
# ==========================================================

return_reasons = [
    "Damaged Product",
    "Wrong Item Delivered",
    "Product Not as Described",
    "Defective Product",
    "Changed Mind",
    "Late Delivery",
    "Size Issue",
    "Quality Not Satisfactory"
]

refund_statuses = [
    "Processed",
    "Processed",
    "Processed",
    "Pending",
    "Rejected"
]

returns = []

return_id = 1

# Only some delivered orders are returned
delivered_orders = orders_df[
    orders_df["OrderStatus"] == "Delivered"
]

for _, order in delivered_orders.iterrows():

    # Approximately 12% return rate
    if random.random() > 0.12:
        continue

    return_date = order["OrderDate"] + timedelta(
        days=random.randint(7, 30)
    )

    refund_amount = round(
        order["TotalAmount"] * random.uniform(0.80, 1.00),
        2
    )

    return_record = {

        "ReturnID": return_id,

        "OrderID": order["OrderID"],

        "CustomerID": order["CustomerID"],

        "ProductID": order["ProductID"],

        "ReturnDate": return_date,

        "ReturnReason": random.choice(return_reasons),

        "RefundAmount": refund_amount,

        "RefundStatus": random.choice(refund_statuses)

    }

    returns.append(return_record)

    return_id += 1

returns_df = pd.DataFrame(returns)

print(returns_df.head())

print("\nShape :", returns_df.shape)

print("\nInfo")
print(returns_df.info())

print("\nNull Values")
print(returns_df.isnull().sum())

print("\nReturn Reason Distribution")
print(returns_df["ReturnReason"].value_counts())

print("\nRefund Status Distribution")
print(returns_df["RefundStatus"].value_counts())

# ==========================================================
# SAVE CSV
# ==========================================================

import os

os.makedirs("../data/raw", exist_ok=True)

returns_df.to_csv(
    "../Data/raw/returns.csv",
    index=False
)

print("\n✅ returns.csv saved successfully!")
```

```python
# ==========================================================
# LINK ORDERS WITH CAMPAIGNS
# ==========================================================

import pandas as pd
import numpy as np

# Load CSV files
orders_df = pd.read_csv("../Data/raw/orders.csv")
campaigns_df = pd.read_csv("../Data/raw/campaigns.csv")

# Convert dates
orders_df["OrderDate"] = pd.to_datetime(orders_df["OrderDate"])
campaigns_df["StartDate"] = pd.to_datetime(campaigns_df["StartDate"])
campaigns_df["EndDate"] = pd.to_datetime(campaigns_df["EndDate"])

# New column
orders_df["CampaignID"] = np.nan

# Assign CampaignID based on campaign duration
for i, order in orders_df.iterrows():

    matching_campaigns = campaigns_df[
        (campaigns_df["StartDate"] <= order["OrderDate"]) &
        (campaigns_df["EndDate"] >= order["OrderDate"])
    ]

    if len(matching_campaigns) > 0:

        selected_campaign = matching_campaigns.sample(1).iloc[0]

        orders_df.at[i, "CampaignID"] = int(selected_campaign["CampaignID"])

# Convert to nullable integer
orders_df["CampaignID"] = orders_df["CampaignID"].astype("Int64")

print(orders_df.head())

print("\nOrders with Campaigns:")
print(orders_df["CampaignID"].notna().sum())

print("\nOrders without Campaign:")
print(orders_df["CampaignID"].isna().sum())

# Save updated Orders file
orders_df.to_csv(
    "../data/raw/orders.csv",
    index=False
)

print("\n✅ orders.csv updated successfully with CampaignID!")
```
